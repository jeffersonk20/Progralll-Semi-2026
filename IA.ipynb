```python
#Librerias a importar
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
#Leer archivo
datos = pd.read_csv('/content/conversor.csv', sep= ';', encoding='utf-8')
print(datos["celsius"])
```

```python
#Obtener datos
c =  datos["celsius"]
f = datos["farenheits"]
print (c,f)
```

```python
#Crear modelo
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
#Compilar el modelo
modelo.compile(optimizer=tf.keras.optimizers.Adam(1.0), loss='mean_squared_error')
```

```python
#Entrenar modelo
epocas = modelo.fit(c, f, epochs=200, verbose=1)
```

```python
resp = modelo.predict(np.array([100.0]))
print(resp)
```
