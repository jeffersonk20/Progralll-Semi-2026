**- Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias las deducciones: ISSS, AFP, ISR (Ver tabla del MH)**

```python
import math

calcular_salario_bruto = lambda hrs, valor_horas: hrs * valor_horas * 30

# Lambda para calcular la deducción de ISSS (3% con tope de $1000 de salario base)
# Tope de deducción ISSS = 1000 * 0.03 = $30.00
calcular_ISSS = lambda salario_bruto: min(salario_bruto * 0.03, 30.00)

# Lambda para calcular la deducción de AFP (7.25% con tope de $7000 de salario base)
# Tope de deducción AFP = 7000 * 0.0725 = $507.50
calcular_AFP = lambda salario_bruto: min(salario_bruto * 0.0725, 507.50)

# Lambda para calcular el ISR (Impuesto Sobre la Renta) para El Salvador
# Basado en la tabla de ISR vigente proporcionada por el usuario
# La renta gravada es el Salario Bruto - ISSS - AFP
calcular_ISR = lambda renta_gravada: \
    0.00 if renta_gravada <= 550.00 else \
    (17.67 + (renta_gravada - 550.00) * 0.10) if renta_gravada <= 895.24 else \
    (60.00 + (renta_gravada - 895.24) * 0.20) if renta_gravada <= 2038.10 else \
    (288.57 + (renta_gravada - 2038.10) * 0.30)
```

```python
import math

# Definir horas trabajadas diarias y valor por hora
horas_trabajadas_diarias = float(input("Ingrese horas trabajadas diarias: "))
valor_por_hora = float(input("Ingrese el valor por hora: "))

print(f"--- Cálculo para {horas_trabajadas_diarias} horas/día a ${valor_por_hora}/hora ---")

# 1. Calcular Salario Bruto
salario_bruto = calcular_salario_bruto(horas_trabajadas_diarias, valor_por_hora)
print(f"Salario Bruto: ${salario_bruto:.2f}")

# 2. Calcular Deducción de ISSS
deduccion_isss = calcular_ISSS(salario_bruto)
print(f"Deducción ISSS: ${deduccion_isss:.2f}")

# 3. Calcular Deducción de AFP
deduccion_afp = calcular_AFP(salario_bruto)
print(f"Deducción AFP: ${deduccion_afp:.2f}")

# 4. Calcular Renta Gravada (Salario Bruto - ISSS - AFP) para el cálculo del ISR
renta_gravada = salario_bruto - deduccion_isss - deduccion_afp

# Aplicar lógica de redondeo específica para la visualización de Renta Gravada
displayed_renta_gravada = math.ceil(renta_gravada * 100) / 100
print(f"Renta Gravada (para ISR): ${displayed_renta_gravada:.2f}")

# 5. Calcular Deducción de ISR
deduccion_isr = calcular_ISR(renta_gravada)
print(f"Deducción ISR: ${deduccion_isr:.6f}")

# Calcular y mostrar el Total de Descuentos
total_descuentos = deduccion_isss + deduccion_afp + deduccion_isr
print(f"Total de Descuentos: ${total_descuentos:.2f}")

# 6. Calcular Salario Neto
salario_neto = salario_bruto - total_descuentos
print(f"Salario Neto: ${salario_neto:.2f}")
```

1. **Calcular** sobre el sueldo las deducciones por ley ISSS (3%), AFP(6.25%), ISR(Según tabla) e imprima el total deducciones y sueldo neto.

```python
while True:
    sueldo = float(input("Ingrese el sueldo: $"))

    # Cálculo de ISSS (3% con tope de $30.00)
    if sueldo * 0.03 > 30.00:
        isss = 30.00
    else:
        isss = sueldo * 0.03

    # Cálculo de AFP (7.25% con tope de $507.50)
    if sueldo * 0.0725 > 507.50:
        afp = 507.50
    else:
        afp = sueldo * 0.0725

    # Calcular la Renta Gravada
    renta_gravada = sueldo - isss - afp

    # Cálculo de ISR usando if/elif/else
    if renta_gravada <= 550.00:
        isr = 0.00
    elif renta_gravada <= 895.24:
        isr = (17.67 + (renta_gravada - 550.00) * 0.10)
    elif renta_gravada <= 2038.10:
        isr = (60.00 + (renta_gravada - 895.24) * 0.20)
    else:
        isr = (288.57 + (renta_gravada - 2038.10) * 0.30)

    total_deducciones = isss + afp + isr

    # Sueldo neto
    sueldo_neto = sueldo - total_deducciones

    print("\n--- DEDUCCIONES ---")
    print(f"Sueldo:             ${sueldo:.2f}")
    print(f"ISSS (3%):          ${isss:.2f}")
    print(f"AFP (7.25%):        ${afp:.2f}")
    print(f"ISR:                ${isr:.2f}")
    print(f"Total deducciones:  ${total_deducciones:.2f}")
    print(f"Sueldo neto:        ${sueldo_neto:.2f}")

    continuar = input("\n¿Desea calcular otro sueldo? (s/n): ").lower()
    if continuar != 's':
        break
print("Fin del cálculo de sueldo.")
```

```python


# ============================================================
#              PROGRAMA COMPLETO DE CONVERSORES
# ============================================================

ejecutando = True


# ============================================================
#              FUNCIÓN GENERAL DE CONVERSIÓN
# ============================================================

def conversor(nombre, unidades, factores):

    while True:

        print("\n========================================")
        print(f"        CONVERSOR DE {nombre.upper()}")
        print("========================================")

        for numero, unidad in unidades.items():
            print(f"{numero}. {unidad}")

        # ----------------------------------------------------
        # Elegir unidad de origen
        # ----------------------------------------------------

        while True:
            try:
                origen = int(input("\n¿De qué unidad desea convertir?: "))

                if origen in unidades:
                    break
                else:
                    print("Opción no válida. Intente nuevamente.")

            except ValueError:
                print("Debe ingresar un número.")

        # ----------------------------------------------------
        # Elegir unidad de destino
        # ----------------------------------------------------

        while True:
            try:
                destino = int(input("¿A qué unidad desea convertir?: "))

                if destino in unidades:
                    break
                else:
                    print("Opción no válida. Intente nuevamente.")

            except ValueError:
                print("Debe ingresar un número.")

        # ----------------------------------------------------
        # Ingresar cantidad
        # ----------------------------------------------------

        while True:
            try:
                valor = float(
                    input("Ingrese el valor que desea convertir: ")
                )

                if valor < 0:
                    print("El valor no puede ser negativo.")
                else:
                    break

            except ValueError:
                print("Debe ingresar un número válido.")

        # ====================================================
        # CONVERSIÓN DE MONEDAS
        # ====================================================

        if nombre == "Monedas":

            resultado = (
                valor * factores[destino]
            ) / factores[origen]

        # ====================================================
        # CONVERSIÓN DE LAS DEMÁS UNIDADES
        # ====================================================

        else:

            unidad_base = valor * factores[origen]

            resultado = unidad_base / factores[destino]

        # ----------------------------------------------------
        # Mostrar resultado
        # ----------------------------------------------------

        print("\n========================================")
        print("        RESULTADO DE LA CONVERSIÓN")
        print("========================================")

        print(
            f"{valor} {unidades[origen]} = "
            f"{resultado:.6f} {unidades[destino]}"
        )

        # ====================================================
        # OPCIONES DESPUÉS DE LA CONVERSIÓN
        # ====================================================

        while True:

            print("\n11. Realizar otra conversión")
            print("12. Salir")
            print("13. Volver al menú principal")

            try:
                continuar = int(
                    input("\nSeleccione una opción: ")
                )

                if continuar == 11:

                    break

                elif continuar == 12:

                    print("\nPrograma finalizado.")
                    return "salir"

                elif continuar == 13:

                    print("\nVolviendo al menú principal...")
                    return "menu"

                else:

                    print(
                        "Opción no válida. "
                        "Intente nuevamente."
                    )

            except ValueError:

                print("Debe ingresar un número.")


# ============================================================
#                    CONVERSOR DE MONEDAS
# ============================================================

monedas = {
    1: "Dólares",
    2: "Euros",
    3: "Libras esterlinas",
    4: "Yenes",
    5: "Dólares canadienses",
    6: "Pesos mexicanos",
    7: "Quetzales",
    8: "Lempiras",
    9: "Colones costarricenses",
    10: "Córdobas"
}


# ============================================================
# TASAS DE MONEDAS
#
# MONEDA DE REFERENCIA: EURO
#
# Los valores representan:
#
# 1 EUR = cantidad de moneda indicada
# ============================================================

factores_monedas = {

    1: 1.1605,
    2: 1.0000,
    3: 0.85608,
    4: 184.62,
    5: 1.6098,
    6: 19.8072,
    7: 8.7516,
    8: 30.7108,
    9: 515.017,
    10: 42.53
}


# ============================================================
#                    CONVERSOR DE LONGITUD
# ============================================================

longitud = {
    1: "Kilómetros",
    2: "Metros",
    3: "Centímetros",
    4: "Milímetros",
    5: "Millas",
    6: "Yardas",
    7: "Pies",
    8: "Pulgadas",
    9: "Millas náuticas",
    10: "Micrómetros"
}

factores_longitud = {
    1: 1000,
    2: 1,
    3: 0.01,
    4: 0.001,
    5: 1609.344,
    6: 0.9144,
    7: 0.3048,
    8: 0.0254,
    9: 1852,
    10: 0.000001
}


# ============================================================
#                      CONVERSOR DE MASA
# ============================================================

masa = {
    1: "Toneladas",
    2: "Kilogramos",
    3: "Hectogramos",
    4: "Decagramos",
    5: "Gramos",
    6: "Decigramos",
    7: "Centigramos",
    8: "Miligramos",
    9: "Libras",
    10: "Onzas"
}

factores_masa = {
    1: 1000000,
    2: 1000,
    3: 100,
    4: 10,
    5: 1,
    6: 0.1,
    7: 0.01,
    8: 0.001,
    9: 453.59237,
    10: 28.349523125
}


# ============================================================
#                 CONVERSOR DE ALMACENAMIENTO
# ============================================================

almacenamiento = {
    1: "Bytes",
    2: "Kilobytes",
    3: "Megabytes",
    4: "Gigabytes",
    5: "Terabytes",
    6: "Petabytes",
    7: "Exabytes",
    8: "Bits",
    9: "Kilobits",
    10: "Megabits"
}

factores_almacenamiento = {
    1: 1,
    2: 1024,
    3: 1024**2,
    4: 1024**3,
    5: 1024**4,
    6: 1024**5,
    7: 1024**6,
    8: 0.125,
    9: 1024 / 8,
    10: (1024**2) / 8
}


# ============================================================
#                    CONVERSOR DE TIEMPO
# ============================================================

tiempo = {
    1: "Microsegundos",
    2: "Milisegundos",
    3: "Segundos",
    4: "Minutos",
    5: "Horas",
    6: "Días",
    7: "Semanas",
    8: "Meses",
    9: "Años",
    10: "Décadas",
    11: "Siglos"
}

factores_tiempo = {
    1: 0.000001,
    2: 0.001,
    3: 1,
    4: 60,
    5: 3600,
    6: 86400,
    7: 604800,
    8: 2592000,
    9: 31536000,
    10: 315360000,
    11: 3153600000
}


# ============================================================
#                    CONVERSOR DE VOLUMEN
# ============================================================

volumen = {
    1: "Litros",
    2: "Mililitros",
    3: "Metros cúbicos",
    4: "Centímetros cúbicos",
    5: "Galones",
    6: "Pintas",
    7: "Cuartos",
    8: "Tazas",
    9: "Onzas líquidas",
    10: "Decilitros"
}

factores_volumen = {
    1: 1,
    2: 0.001,
    3: 1000,
    4: 0.001,
    5: 3.785411784,
    6: 0.473176473,
    7: 0.946352946,
    8: 0.2365882365,
    9: 0.0295735296,
    10: 0.1
}


# ============================================================
#                      CONVERSOR DE ÁREA
# ============================================================

area = {
    1: "Kilómetros cuadrados",
    2: "Hectáreas",
    3: "Metros cuadrados",
    4: "Decímetros cuadrados",
    5: "Centímetros cuadrados",
    6: "Milímetros cuadrados",
    7: "Millas cuadradas",
    8: "Yardas cuadradas",
    9: "Pies cuadrados",
    10: "Acres"
}

factores_area = {
    1: 1000000,
    2: 10000,
    3: 1,
    4: 0.01,
    5: 0.0001,
    6: 0.000001,
    7: 2589988.110336,
    8: 0.83612736,
    9: 0.09290304,
    10: 4046.8564224
}


# ============================================================
#                     MENÚ PRINCIPAL
# ============================================================

while ejecutando:

    print("\n")
    print("========================================")
    print("           MENÚ PRINCIPAL")
    print("========================================")
    print("1. Monedas")
    print("2. Longitud")
    print("3. Masa")
    print("4. Almacenamiento")
    print("5. Tiempo")
    print("6. Volumen")
    print("7. Área")
    print("8. Salir")
    print("========================================")

    try:

        opcion_principal = int(
            input(
                "\nSeleccione el conversor "
                "que desea utilizar: "
            )
        )

    except ValueError:

        print("\nDebe ingresar un número.")
        continue


    # ========================================================
    # MONEDAS
    # ========================================================

    if opcion_principal == 1:

        resultado = conversor(
            "Monedas",
            monedas,
            factores_monedas
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # LONGITUD
    # ========================================================

    elif opcion_principal == 2:

        resultado = conversor(
            "Longitud",
            longitud,
            factores_longitud
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # MASA
    # ========================================================

    elif opcion_principal == 3:

        resultado = conversor(
            "Masa",
            masa,
            factores_masa
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # ALMACENAMIENTO
    # ========================================================

    elif opcion_principal == 4:

        resultado = conversor(
            "Almacenamiento",
            almacenamiento,
            factores_almacenamiento
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # TIEMPO
    # ========================================================

    elif opcion_principal == 5:

        resultado = conversor(
            "Tiempo",
            tiempo,
            factores_tiempo
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # VOLUMEN
    # ========================================================

    elif opcion_principal == 6:

        resultado = conversor(
            "Volumen",
            volumen,
            factores_volumen
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # ÁREA
    # ========================================================

    elif opcion_principal == 7:

        resultado = conversor(
            "Área",
            area,
            factores_area
        )

        if resultado == "salir":
            ejecutando = False


    # ========================================================
    # SALIR
    # ========================================================

    elif opcion_principal == 8:

        print("\nPrograma finalizado.")
        ejecutando = False


    # ========================================================
    # OPCIÓN NO VÁLIDA
    # ========================================================

    else:

        print("\nOpción no válida. Intente nuevamente.")


# ============================================================
#                       FIN DEL PROGRAMA
# ============================================================

print("\nGracias por utilizar el programa.")
```

5. **Realizar el sistema de gestión del tiempo y la demanda en las unidades de salud.**

```python
import datetime

def evaluar_prioridad(edad, motivo):
    motivo_lower = motivo.lower()

    # PRIORIDAD 1: Emergencias vitales o signos de gravedad agudos
    rojo_directo = [
        "inconsciente", "paro", "shock", "choque", "hemorragia masiva",
        "politraumatismo", "no respira", "convulsion", "paro cardiaco", "infarto"
    ]
    if any(kw in motivo_lower for kw in rojo_directo):
        return "Prioridad 1"

    sospecha_emergencia = ["dolor de pecho", "dificultad para respirar", "falta de aire", "asfixia"]
    if any(kw in motivo_lower for kw in sospecha_emergencia):
        resp = input("  [?] ¿El paciente presenta dificultad respiratoria severa o dolor de pecho agudo? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    if edad < 5 and any(kw in motivo_lower for kw in ["fiebre", "tos", "diarrea", "vomito", "grave"]):
        resp = input("  [?] ¿El menor presenta letargia, convulsiones o incapacidad para beber/beber pecho? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    if any(kw in motivo_lower for kw in ["embarazo", "embarazada", "parto", "puerpera", "sangrado vaginal"]):
        resp = input("  [?] ¿Presenta sangrado vaginal, convulsiones o fuerte dolor abdominal? (s/n): ").strip().lower()
        if resp == 's':
            return "Prioridad 1"

    # PRIORIDAD 2: Vulnerables o casos agudos/descompensados
    if edad >= 65 or edad < 1:
        return "Prioridad 2"

    if any(kw in motivo_lower for kw in ["embarazo", "embarazada", "puerpera"]):
        return "Prioridad 2"

    p2_keywords = [
        "descompensad", "diabetes", "hipertension", "presion alta", "presion baja",
        "dolor fuerte", "dolor moderado", "fractura", "quemadura", "alergia", "asma"
    ]
    if any(kw in motivo_lower for kw in p2_keywords):
        return "Prioridad 2"

    # PRIORIDAD 3: Consultas leves o rutinarias
    return "Prioridad 3"


def sistema_triage():
    pacientes = []
    orden_prioridad = {"Prioridad 1": 1, "Prioridad 2": 2, "Prioridad 3": 3}
    contador_llegada = 0  # Garantiza el orden de llegada exacto

    while True:
        print("\n--- MENÚ PRINCIPAL ---")
        print("1. Agregar paciente")
        print("2. Ver lista de pacientes por orden de atención")
        print("3. Salir")
        opcion = input("Seleccione una opción: ").strip()

        if opcion == '1':
            print("\n--- NUEVO PACIENTE ---")
            nombre = input("Nombre del paciente: ").strip()

            while True:
                try:
                    edad = int(input("Edad: "))
                    if edad >= 0:
                        break
                    print("Ingrese una edad válida.")
                except ValueError:
                    print("Por favor, ingrese un número entero.")

            motivo = input("Motivo de consulta: ").strip()

            prioridad = evaluar_prioridad(edad, motivo)
            contador_llegada += 1  # Asigna turno correlativo según llegada

            pacientes.append({
                "nombre": nombre,
                "motivo": motivo,
                "prioridad": prioridad,
                "orden": orden_prioridad[prioridad],
                "llegada": contador_llegada
            })

            print(f"\n[OK] Paciente {nombre.upper()} asignado a: {prioridad}")

            sub_opcion = input("\n¿Desea agregar otro paciente? (s/n): ").strip().lower()
            if sub_opcion != 's':
                pass

        elif opcion == '2':
            print("\n--- LISTA DE PACIENTES (ORDEN DE ATENCIÓN) ---")
            if not pacientes:
                print("No hay pacientes registrados.")
            else:
                # Se ordena PRIMERO por 'orden' (Prioridad 1, 2, 3)
                # y SEGUNDO por 'llegada' (el que ingresó antes pasa primero)
                pacientes_ordenados = sorted(pacientes, key=lambda x: (x['orden'], x['llegada']))

                for i, p in enumerate(pacientes_ordenados, 1):
                    print(f"{i}. {p['nombre']} - {p['prioridad']} | Motivo: {p['motivo']}")

        elif opcion == '3':
            print("\nSaliendo del programa...")
            break
        else:
            print("Opción no válida. Intente nuevamente.")

if __name__ == "__main__":
    sistema_triage()
```
