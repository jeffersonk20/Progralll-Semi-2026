```python
print("Sumar dos numeros: ")
num1 = input("NUM 1: ")
num2 = input("Num 2: ")
respuesta = float(num1) + float(num2)

print(type(num1), type(num2), type(respuesta))

print("Respuesta: ", respuesta)


print("Suma completa")
print(respuesta, type(respuesta))
```

```python
print("Restar dos numeros: ")
num1 = input("NUM 1: ")
num2 = input("Num 2: ")
respuesta = float(num1) - float(num2)

print(type(num1), type(num2), type(respuesta))

print("Respuesta: ", respuesta)
```

```python
print("Multiplicar dos numeros: ")
num1 = input("NUM 1: ")
num2 = input("Num 2: ")
respuesta = float(num1) * float(num2)

print(type(num1), type(num2), type(respuesta))

print("Respuesta: ", respuesta)


print("Multiplicacion completa")
print(respuesta, type(respuesta))
```
