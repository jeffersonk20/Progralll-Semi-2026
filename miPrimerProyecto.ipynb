**- Calcular el sueldo de un empleado. hrs * valor de horas * 30 dias las deducciones: ISSS, AFP, ISR (Ver tabla del MH)**

```python
import math

calcular_salario_bruto = lambda hrs, valor_horas: hrs * valor_horas * 30

# Lambda para calcular la deducción de ISSS (3% con tope de $1000 de salario base)
# Tope de deducción ISSS = 1000 * 0.03 = $30.00
calcular_ISSS = lambda salario_bruto: min(salario_bruto * 0.03, 30.00)

# Lambda para calcular la deducción de AFP (7.25% con tope de $7000 de salario base)
# Tope de deducción AFP = 7000 * 0.0725 = $507.50
calcular_AFP = lambda salario_bruto: min(salario_bruto * 0.0725, 507.50)

# Lambda para calcular el ISR (Impuesto Sobre la Renta) para El Salvador
# Basado en la tabla de ISR vigente proporcionada por el usuario
# La renta gravada es el Salario Bruto - ISSS - AFP
calcular_ISR = lambda renta_gravada: \
    0.00 if renta_gravada <= 550.00 else \
    (17.67 + (renta_gravada - 550.00) * 0.10) if renta_gravada <= 895.24 else \
    (60.00 + (renta_gravada - 895.24) * 0.20) if renta_gravada <= 2038.10 else \
    (288.57 + (renta_gravada - 2038.10) * 0.30)
```

```python
import math

# Definir horas trabajadas diarias y valor por hora
horas_trabajadas_diarias = float(input("Ingrese horas trabajadas diarias: "))
valor_por_hora = float(input("Ingrese el valor por hora: "))

print(f"--- Cálculo para {horas_trabajadas_diarias} horas/día a ${valor_por_hora}/hora ---")

# 1. Calcular Salario Bruto
salario_bruto = calcular_salario_bruto(horas_trabajadas_diarias, valor_por_hora)
print(f"Salario Bruto: ${salario_bruto:.2f}")

# 2. Calcular Deducción de ISSS
deduccion_isss = calcular_ISSS(salario_bruto)
print(f"Deducción ISSS: ${deduccion_isss:.2f}")

# 3. Calcular Deducción de AFP
deduccion_afp = calcular_AFP(salario_bruto)
print(f"Deducción AFP: ${deduccion_afp:.2f}")

# 4. Calcular Renta Gravada (Salario Bruto - ISSS - AFP) para el cálculo del ISR
renta_gravada = salario_bruto - deduccion_isss - deduccion_afp

# Aplicar lógica de redondeo específica para la visualización de Renta Gravada
displayed_renta_gravada = math.ceil(renta_gravada * 100) / 100
print(f"Renta Gravada (para ISR): ${displayed_renta_gravada:.2f}")

# 5. Calcular Deducción de ISR
deduccion_isr = calcular_ISR(renta_gravada)
print(f"Deducción ISR: ${deduccion_isr:.6f}")

# Calcular y mostrar el Total de Descuentos
total_descuentos = deduccion_isss + deduccion_afp + deduccion_isr
print(f"Total de Descuentos: ${total_descuentos:.2f}")

# 6. Calcular Salario Neto
salario_neto = salario_bruto - total_descuentos
print(f"Salario Neto: ${salario_neto:.2f}")
```

1. **Calcular** sobre el sueldo las deducciones por ley ISSS (3%), AFP(6.25%), ISR(Según tabla) e imprima el total deducciones y sueldo neto.

```python
sueldo = float(input("Ingrese el sueldo: $"))

# Cálculo de ISSS (3% con tope de $30.00)
if sueldo * 0.03 > 30.00:
    isss = 30.00
else:
    isss = sueldo * 0.03

# Cálculo de AFP (7.25% con tope de $507.50)
if sueldo * 0.0725 > 507.50:
    afp = 507.50
else:
    afp = sueldo * 0.0725

# Calcular la Renta Gravada
renta_gravada = sueldo - isss - afp

# Cálculo de ISR usando if/elif/else
if renta_gravada <= 550.00:
    isr = 0.00
elif renta_gravada <= 895.24:
    isr = (17.67 + (renta_gravada - 550.00) * 0.10)
elif renta_gravada <= 2038.10:
    isr = (60.00 + (renta_gravada - 895.24) * 0.20)
else:
    isr = (288.57 + (renta_gravada - 2038.10) * 0.30)

total_deducciones = isss + afp + isr

# Sueldo neto
sueldo_neto = sueldo - total_deducciones

print("\n--- DEDUCCIONES ---")
print(f"Sueldo:             ${sueldo:.2f}")
print(f"ISSS (3%):          ${isss:.2f}")
print(f"AFP (7.25%):        ${afp:.2f}")
print(f"ISR:                ${isr:.2f}")
print(f"Total deducciones:  ${total_deducciones:.2f}")
print(f"Sueldo neto:        ${sueldo_neto:.2f}")
```

```python
ejecutando = True

while ejecutando:

    print("\n=== CONVERSOR DE MASA ===")
    print("1. Kilogramos → Gramos")
    print("2. Gramos → Kilogramos")
    print("3. Kilogramos → Libras")
    print("4. Libras → Kilogramos")
    print("5. Gramos → Miligramos")
    print("6. Miligramos → Gramos")
    print("7. Toneladas → Kilogramos")
    print("8. Kilogramos → Toneladas")
    print("9. Libras → Onzas")
    print("10. Onzas → Libras")

    opcion = int(input("\nSeleccione una opción: "))

    if opcion >= 1 and opcion <= 10:

        valor = float(input("Ingrese el valor a convertir: "))

        if opcion == 1:
            resultado = valor * 1000
            print(f"\nResultado: {resultado} gramos")

        elif opcion == 2:
            resultado = valor / 1000
            print(f"\nResultado: {resultado} kilogramos")

        elif opcion == 3:
            resultado = valor * 2.20462
            print(f"\nResultado: {resultado} libras")

        elif opcion == 4:
            resultado = valor / 2.20462
            print(f"\nResultado: {resultado} kilogramos")

        elif opcion == 5:
            resultado = valor * 1000
            print(f"\nResultado: {resultado} miligramos")

        elif opcion == 6:
            resultado = valor / 1000
            print(f"\nResultado: {resultado} gramos")

        elif opcion == 7:
            resultado = valor * 1000
            print(f"\nResultado: {resultado} kilogramos")

        elif opcion == 8:
            resultado = valor / 1000
            print(f"\nResultado: {resultado} toneladas")

        elif opcion == 9:
            resultado = valor * 16
            print(f"\nResultado: {resultado} onzas")

        elif opcion == 10:
            resultado = valor / 16
            print(f"\nResultado: {resultado} libras")

        # Opciones después de realizar la conversión
        while True:

            print("\n11. Realizar otra conversión")
            print("12. Salir")

            continuar = int(input("\nSeleccione una opción: "))

            if continuar == 11:
                break

            elif continuar == 12:
                print("\nPrograma finalizado.")
                ejecutando = False
                break

            else:
                print("\n❌ Opción no válida. Intente nuevamente.")

    else:
        print("\n❌ Opción no válida. Intente nuevamente.")
```
