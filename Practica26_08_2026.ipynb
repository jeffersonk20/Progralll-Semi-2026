```python
class PalindromoChecker:
  def es_palindromo(self, palabra):
    palabra_limpia = ''.join(filter(str.isalnum, palabra)).lower()
    return palabra_limpia == palabra_limpia[::-1]

# Ejemplo para Tarea 3: Palíndromo
print("\n--- Tarea 3: Verificador de Palíndromos ---")
obj_palindromo = PalindromoChecker()

palabra_usuario = input("Ingrese una palabra para verificar si es palíndromo: ")

if obj_palindromo.es_palindromo(palabra_usuario):
  print(f"'{palabra_usuario}' ES un palíndromo.")
else:
  print(f"'{palabra_usuario}' NO es un palíndromo.")
```

```python

from functools import reduce
from collections import Counter
import json


# ==========================================================
# CLASE SISTEMA ACADEMICO
# ==========================================================

class SistemaAcademico:

    def __init__(self):
        self.estudiantes = []

    # ------------------------------------------
    # Agregar estudiante
    # ------------------------------------------

    def agregar_estudiante(self, estudiante):
        self.estudiantes.append(estudiante)

    # ------------------------------------------
    # Mostrar estudiantes
    # ------------------------------------------

    def mostrar_estudiantes(self):

        print("\n==============================================")
        print("          LISTA DE ESTUDIANTES")
        print("==============================================")

        for i, estudiante in enumerate(self.estudiantes, 1):

            print("\nEstudiante:", i)
            print("Nombre:", estudiante["nombre"])

            print("Materias:")

            for materia, nota in estudiante["materias"].items():

                print("  ", materia, ":", nota)

            print("Edad:", estudiante["edad"], "años")

            print(
                "Tiempo de traslado:",
                estudiante["tiempo_traslado"],
                "minutos"
            )

            print(
                "Horas de uso del telefono:",
                estudiante["horas_telefono"],
                "horas"
            )

    # ------------------------------------------
    # Guardar en JSON
    # ------------------------------------------

    def guardar_json(self):

        with open(
            "estudiantes.json",
            "w",
            encoding="utf-8"
        ) as archivo:

            json.dump(
                self.estudiantes,
                archivo,
                indent=4,
                ensure_ascii=False
            )

        print("\nLos datos fueron guardados en estudiantes.json")

    # ------------------------------------------
    # Cargar JSON
    # ------------------------------------------

    def cargar_json(self):

        try:

            with open(
                "estudiantes.json",
                "r",
                encoding="utf-8"
            ) as archivo:

                self.estudiantes = json.load(archivo)

            print("\nDatos cargados correctamente.")

        except FileNotFoundError:

            print("\nNo existe el archivo estudiantes.json.")


# ==========================================================
# CLASE ESTADISTICA
# ==========================================================

class estadistica:

    # ------------------------------------------
    # MEDIA
    # ------------------------------------------

    def media(self, serie):

        suma = reduce(
            lambda acum, num:
            acum + float(num),
            serie,
            0
        )

        return suma / len(serie)

    # ------------------------------------------
    # MEDIANA
    # ------------------------------------------

    def mediana(self, serie):

        datos = sorted(
            [float(num) for num in serie]
        )

        cantidad = len(datos)

        if cantidad % 2 == 0:

            posicion1 = cantidad // 2 - 1
            posicion2 = cantidad // 2

            return (
                datos[posicion1] +
                datos[posicion2]
            ) / 2

        else:

            return datos[cantidad // 2]

    # ------------------------------------------
    # MODA
    # ------------------------------------------

    def moda(self, serie):

        datos = [
            float(num)
            for num in serie
        ]

        frecuencias = Counter(datos)

        mayor = max(
            frecuencias.values()
        )

        if mayor == 1:

            return "No existe moda"

        modas = []

        for numero, frecuencia in frecuencias.items():

            if frecuencia == mayor:

                modas.append(numero)

        return modas

    # ------------------------------------------
    # VARIANZA
    # ------------------------------------------

    def varianza(self, serie):

        media = self.media(serie)

        suma = reduce(
            lambda acum, num:
            acum +
            (float(num) - media) ** 2,
            serie,
            0
        )

        return suma / len(serie)

    # ------------------------------------------
    # DESVIACION ESTANDAR
    # ------------------------------------------

    def desviacion_estandar(self, serie):

        return self.varianza(serie) ** 0.5

    # ------------------------------------------
    # DESVIACION TIPICA
    # ------------------------------------------

    def desviacion_tipica(self, serie):

        return self.desviacion_estandar(serie)

    # ------------------------------------------
    # RANGO
    # ------------------------------------------

    def rango(self, serie):

        datos = [
            float(num)
            for num in serie
        ]

        return max(datos) - min(datos)

    # ------------------------------------------
    # FRECUENCIA
    # ------------------------------------------

    def frecuencia(self, serie):

        datos = [
            float(num)
            for num in serie
        ]

        return dict(
            Counter(datos)
        )

    # ------------------------------------------
    # MOSTRAR TODAS
    # ------------------------------------------

    def mostrar_todas(self, serie):

        print("\n================================")
        print("       RESULTADOS ESTADISTICOS")
        print("================================")

        print(
            "Media:",
            round(self.media(serie), 2)
        )

        print(
            "Mediana:",
            round(self.mediana(serie), 2)
        )

        print(
            "Moda:",
            self.moda(serie)
        )

        print(
            "Varianza:",
            round(self.varianza(serie), 2)
        )

        print(
            "Desviacion estandar:",
            round(
                self.desviacion_estandar(serie),
                2
            )
        )

        print(
            "Desviacion tipica:",
            round(
                self.desviacion_tipica(serie),
                2
            )
        )

        print(
            "Rango:",
            round(self.rango(serie), 2)
        )

        print(
            "Frecuencia:",
            self.frecuencia(serie)
        )


# ==========================================================
# FUNCION PARA MOSTRAR ESTADISTICAS
# ==========================================================

def menu_medidas(objEstadistica, serie, nombre):

    while True:

        print("\n================================")
        print("     ESTADISTICAS DE", nombre)
        print("================================")

        print("1. Media aritmetica")
        print("2. Mediana")
        print("3. Moda")
        print("4. Varianza")
        print("5. Desviacion estandar")
        print("6. Desviacion tipica")
        print("7. Rango")
        print("8. Frecuencia")
        print("9. Mostrar todas")
        print("10. Regresar")

        opcion = input("\nSeleccione una opcion: ")

        if opcion == "1":

            print(
                "\nMedia:",
                round(
                    objEstadistica.media(serie),
                    2
                )
            )

        elif opcion == "2":

            print(
                "\nMediana:",
                round(
                    objEstadistica.mediana(serie),
                    2
                )
            )

        elif opcion == "3":

            print(
                "\nModa:",
                objEstadistica.moda(serie)
            )

        elif opcion == "4":

            print(
                "\nVarianza:",
                round(
                    objEstadistica.varianza(serie),
                    2
                )
            )

        elif opcion == "5":

            print(
                "\nDesviacion estandar:",
                round(
                    objEstadistica.desviacion_estandar(serie),
                    2
                )
            )

        elif opcion == "6":

            print(
                "\nDesviacion tipica:",
                round(
                    objEstadistica.desviacion_tipica(serie),
                    2
                )
            )

        elif opcion == "7":

            print(
                "\nRango:",
                round(
                    objEstadistica.rango(serie),
                    2
                )
            )

        elif opcion == "8":

            print(
                "\nFrecuencia:",
                objEstadistica.frecuencia(serie)
            )

        elif opcion == "9":

            objEstadistica.mostrar_todas(
                serie
            )

        elif opcion == "10":

            break

        else:

            print("\nOpcion no valida.")


# ==========================================================
# MENU ESTADISTICO
# ==========================================================

def menu_estadistico(
    objEstadistica,
    estudiantes
):

    while True:

        print("\n========================================")
        print("          SISTEMA ESTADISTICO")
        print("========================================")

        print("1. Estadisticas de notas")
        print("2. Estadisticas de edades")
        print("3. Estadisticas de tiempo de traslado")
        print("4. Estadisticas de horas de telefono")
        print("5. Regresar")

        opcion = input(
            "\nSeleccione una opcion: "
        )

        # --------------------------------------
        # NOTAS
        # --------------------------------------

        if opcion == "1":

            while True:

                print("\n================================")
                print("       ESTADISTICAS DE NOTAS")
                print("================================")

                print("1. Estadistica")
                print("2. Diseño de Circuitos Digitales")
                print("3. Ingles Basico")
                print("4. Base de Datos II")
                print("5. Programacion Computacional III")
                print("6. Regresar")

                opcion_nota = input(
                    "\nSeleccione una materia: "
                )

                materias = []

                if opcion_nota == "1":

                    materia_buscada = "Estadistica"

                elif opcion_nota == "2":

                    materia_buscada = (
                        "Diseño de Circuitos Digitales"
                    )

                elif opcion_nota == "3":

                    materia_buscada = "Ingles Basico"

                elif opcion_nota == "4":

                    materia_buscada = "Base de Datos II"

                elif opcion_nota == "5":

                    materia_buscada = (
                        "Programacion Computacional III"
                    )

                elif opcion_nota == "6":

                    break

                else:

                    print("\nOpcion no valida.")
                    continue

                # Obtener las notas
                for estudiante in estudiantes:

                    if materia_buscada in estudiante["materias"]:

                        materias.append(
                            estudiante["materias"][
                                materia_buscada
                            ]
                        )

                if len(materias) > 0:

                    menu_medidas(
                        objEstadistica,
                        materias,
                        materia_buscada
                    )

        # --------------------------------------
        # EDADES
        # --------------------------------------

        elif opcion == "2":

            serie = []

            for estudiante in estudiantes:

                serie.append(
                    estudiante["edad"]
                )

            menu_medidas(
                objEstadistica,
                serie,
                "EDADES"
            )

        # --------------------------------------
        # TRASLADO
        # --------------------------------------

        elif opcion == "3":

            serie = []

            for estudiante in estudiantes:

                serie.append(
                    estudiante[
                        "tiempo_traslado"
                    ]
                )

            menu_medidas(
                objEstadistica,
                serie,
                "TIEMPO DE TRASLADO"
            )

        # --------------------------------------
        # TELEFONO
        # --------------------------------------

        elif opcion == "4":

            serie = []

            for estudiante in estudiantes:

                serie.append(
                    estudiante[
                        "horas_telefono"
                    ]
                )

            menu_medidas(
                objEstadistica,
                serie,
                "HORAS DE TELEFONO"
            )

        elif opcion == "5":

            break

        else:

            print("\nOpcion no valida.")


# ==========================================================
# CREAR OBJETOS
# ==========================================================

objAcademico = SistemaAcademico()

objEstadistica = estadistica()


# ==========================================================
# 20 ESTUDIANTES
# ==========================================================

objAcademico.agregar_estudiante({

    "nombre":
    "SAMUEL ALEXANDER PORTILLO CERÓN",

    "materias": {

        "Estadistica": 10,

        "Diseño de Circuitos Digitales": 10,

        "Ingles Basico": 10,

        "Base de Datos II": 10

    },

    "edad": 20,

    "tiempo_traslado": 80,

    "horas_telefono": 6
})


objAcademico.agregar_estudiante({

    "nombre":
    "CARLOS DAVID TURCIOS RIVAS",

    "materias": {

        "Estadistica": 9,

        "Diseño de Circuitos Digitales": 9,

        "Ingles Basico": 10,

        "Base de Datos II": 10

    },

    "edad": 20,

    "tiempo_traslado": 25,

    "horas_telefono": 5
})


objAcademico.agregar_estudiante({

    "nombre":
    "Gilda Mileidy González Castro",

    "materias": {

        "Estadistica": 7.5,

        "Diseño de Circuitos Digitales": 10,

        "Ingles Basico": 10,

        "Base de Datos II": 10

    },

    "edad": 19,

    "tiempo_traslado": 30,

    "horas_telefono": 1.33
})


objAcademico.agregar_estudiante({

    "nombre":
    "Kimberly Michelle Castro Zealaya",

    "materias": {

        "Programacion Computacional III": 10

    },

    "edad": 18,

    "tiempo_traslado": 20,

    "horas_telefono": 9
})


objAcademico.agregar_estudiante({

    "nombre":
    "Yeimy Julissa Parada Chávez",

    "materias": {

        "Base de Datos II": 9.5,

        "Ingles Basico": 9.5,

        "Programacion Computacional III": 9.5,

        "Diseño de Circuitos Digitales": 9.5,

        "Estadistica": 9.5

    },

    "edad": 20,

    "tiempo_traslado": 45,

    "horas_telefono": 3
})


objAcademico.agregar_estudiante({

    "nombre":
    "SAÚL ALEXANDER JIMENEZ GARCIA",

    "materias": {

        "Programacion Computacional II": 10

    },

    "edad": 21,

    "tiempo_traslado": 60,

    "horas_telefono": 5
})


objAcademico.agregar_estudiante({

    "nombre":
    "Josué Stiven Funes Benavides",

    "materias": {

        "Base de Datos II": 9.5,

        "Ingles Basico": 9.5,

        "Programacion Computacional III": 9.5,

        "Diseño de Circuitos Digitales": 9.5,

        "Estadistica": 9.5

    },

    "edad": 19,

    "tiempo_traslado": 30,

    "horas_telefono": 5
})


objAcademico.agregar_estudiante({

    "nombre":
    "JEFFERSON STEVEN CASTILLO LEIVA",

    "materias": {

        "Base de Datos II": 9.5,

        "Ingles Basico": 9.5,

        "Programacion Computacional III": 9.5,

        "Diseño de Circuitos Digitales": 9.5,

        "Estadistica": 9.5

    },

    "edad": 19,

    "tiempo_traslado": 15,

    "horas_telefono": 2.30
})


objAcademico.agregar_estudiante({

    "nombre":
    "RAMAEL ERNESTO CORDOVA COLINDRES",

    "materias": {

        "Base de Datos II": 9.5,

        "Ingles Basico": 9.5,

        "Programacion Computacional III": 9.5,

        "Diseño de Circuitos Digitales": 9.5,

        "Estadistica": 9.5

    },

    "edad": 22,

    "tiempo_traslado": 7,

    "horas_telefono": 13.52
})


objAcademico.agregar_estudiante({

    "nombre":
    "Carlos Daniel Parada Avalos",

    "materias": {

        "Base de Datos II": 8.5,

        "Ingles Basico": 8.5,

        "Programacion Computacional III": 8.5,

        "Diseño de Circuitos Digitales": 8.5,

        "Estadistica": 8.5

    },

    "edad": 25,

    "tiempo_traslado": 10,

    "horas_telefono": 5.52
})


objAcademico.agregar_estudiante({

    "nombre":
    "Karen Elizabet Ulloa Gómez",

    "materias": {

        "Base de Datos II": 9,

        "Ingles Basico": 9,

        "Programacion Computacional III": 9,

        "Diseño de Circuitos Digitales": 9,

        "Estadistica": 9

    },

    "edad": 18,

    "tiempo_traslado": 40,

    "horas_telefono": 8
})


objAcademico.agregar_estudiante({

    "nombre":
    "Madison Odalis Valle Juárez",

    "materias": {

        "Base de Datos II": 9,

        "Ingles Basico": 9,

        "Programacion Computacional III": 9,

        "Diseño de Circuitos Digitales": 9,

        "Estadistica": 9

    },

    "edad": 20,

    "tiempo_traslado": 40,

    "horas_telefono": 9.33
})


objAcademico.agregar_estudiante({

    "nombre":
    "Miguel Alejandro Gómez Saravia",

    "materias": {

        "Base de Datos II": 9.5,

        "Ingles Basico": 9.5,

        "Programacion Computacional III": 9.5,

        "Diseño de Circuitos Digitales": 9.5,

        "Estadistica": 9.5

    },

    "edad": 19,

    "tiempo_traslado": 15,

    "horas_telefono": 4
})


objAcademico.agregar_estudiante({

    "nombre":
    "Nehemias Isaí Martínez Argueta",

    "materias": {

        "Base de Datos II": 8,

        "Ingles Basico": 8,

        "Programacion Computacional III": 8,

        "Diseño de Circuitos Digitales": 8,

        "Estadistica": 8

    },

    "edad": 19,

    "tiempo_traslado": 45,

    "horas_telefono": 9
})


objAcademico.agregar_estudiante({

    "nombre":
    "Sharon Nicol Argüello Torres",

    "materias": {

        "Programacion Computacional III": 8

    },

    "edad": 20,

    "tiempo_traslado": 20,

    "horas_telefono": 6
})


objAcademico.agregar_estudiante({

    "nombre":
    "Jenifer Maricela Tenderos Guevara",

    "materias": {

        "Base de Datos II": 9.5,

        "Ingles Basico": 9.5,

        "Programacion Computacional III": 9.5,

        "Diseño de Circuitos Digitales": 9.5,

        "Estadistica": 9.5

    },

    "edad": 21,

    "tiempo_traslado": 60,

    "horas_telefono": 3
})


objAcademico.agregar_estudiante({

    "nombre":
    "Claudia Arely Sorto Meléndez",

    "materias": {

        "Base de Datos II": 8.5,

        "Ingles Basico": 8.5,

        "Programacion Computacional III": 8.5,

        "Diseño de Circuitos Digitales": 8.5,

        "Estadistica": 8.5

    },

    "edad": 19,

    "tiempo_traslado": 50,

    "horas_telefono": 7.62
})


objAcademico.agregar_estudiante({

    "nombre":
    "Daniel Enrique Quintanilla Paniagua",

    "materias": {

        "Base de Datos II": 9.5,

        "Ingles Basico": 9.5,

        "Programacion Computacional III": 9.5,

        "Diseño de Circuitos Digitales": 9.5,

        "Estadistica": 9.5

    },

    "edad": 19,

    "tiempo_traslado": 15,

    "horas_telefono": 4
})


objAcademico.agregar_estudiante({

    "nombre":
    "José Roberto Díaz",

    "materias": {

        "Base de Datos II": 9,

        "Ingles Basico": 9,

        "Programacion Computacional III": 9,

        "Diseño de Circuitos Digitales": 9,

        "Estadistica": 9

    },

    "edad": 21,

    "tiempo_traslado": 75,

    "horas_telefono": 5
})


objAcademico.agregar_estudiante({

    "nombre":
    "Kevin Manrique Campos",

    "materias": {

        "Base de Datos II": 9.5,

        "Ingles Basico": 9.5,

        "Programacion Computacional III": 9.5,

        "Diseño de Circuitos Digitales": 9.5,

        "Estadistica": 9.5

    },

    "edad": 19,

    "tiempo_traslado": 30,

    "horas_telefono": 6
})


# ==========================================================
# MENU PRINCIPAL
# ==========================================================

while True:

    print("\n")
    print("========================================")
    print("         SISTEMA ACADEMICO UGB")
    print("========================================")

    print("1. Mostrar estudiantes")
    print("2. Guardar datos en JSON")
    print("3. Cargar datos desde JSON")
    print("4. Sistema estadistico")
    print("5. Salir")

    opcion = input(
        "\nSeleccione una opcion: "
    )

    if opcion == "1":

        objAcademico.mostrar_estudiantes()

    elif opcion == "2":

        objAcademico.guardar_json()

    elif opcion == "3":

        objAcademico.cargar_json()

    elif opcion == "4":

        menu_estadistico(
            objEstadistica,
            objAcademico.estudiantes
        )

    elif opcion == "5":

        print("\nPrograma finalizado.")
        break

    else:

        print("\nOpcion no valida.")
```
